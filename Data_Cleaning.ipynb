```python
import pandas as pd

df = pd.read_csv("marketing_campaign.csv", sep=None, engine="python")

print("Rows and Columns:", df.shape)
```

```python
print("Column Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Show rows where Income is missing
missing_income = df[df["Income"].isnull()]

print("Number of missing Income values:", missing_income.shape[0])

display(missing_income)
```

```python
print("Income Statistics:")
print(df["Income"].describe())
```

```python
print("Income Median:", df["Income"].median())
print("Income Mean:", df["Income"].mean())
```

```python
# Fill missing Income values with the median
income_median = df["Income"].median()

df["Income"] = df["Income"].fillna(income_median)

print("Missing Income values after cleaning:")
print(df["Income"].isnull().sum())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Duplicate rows before cleaning:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
print("Final dataset shape:", df.shape)
```

```python
print("Education values:")
print(df["Education"].value_counts())
```

```python
text_columns = df.select_dtypes(include="object").columns

print("Text columns:")
print(text_columns.tolist())

for col in text_columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts().head(20))
```

```python
# Remove leading/trailing spaces from text columns
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()

print("Extra spaces removed from text columns.")
```

```python
df["Education"] = df["Education"].replace({
    "2n Cycle": "2nd Cycle"
})

print(df["Education"].value_counts())
```

```python
df["Marital_Status"] = df["Marital_Status"].str.title()

print(df["Marital_Status"].value_counts())
```

```python
df['customer_year'] = df['dt_customer'].dt.year
```

```python
df['customer_month_name'] = df['dt_customer'].dt.month_name()
```

```python
reference_date = df['dt_customer'].max()

df['customer_tenure_years'] = (
    (reference_date - df['dt_customer']).dt.days / 365.25
).round(2)
```
